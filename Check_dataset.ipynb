```python
import os
import pandas as pd
from google.colab import drive

# Mount Google Drive
drive.mount('/content/drive')

# Define the base path - YOUR ACTUAL PATH
BASE_PATH = '/content/drive/MyDrive/Colab Notebooks/New'

# CORRECT file path based on your folder structure
CSV_FILE_PATH = os.path.join(BASE_PATH, 'ExpTOR', 'Preprocessed', 'Physio', '90s_features.csv')

INDICATORS_PATH = os.path.join(BASE_PATH, 'physiological_indicators.xlsx')

print("🔍 Checking file paths...")
print(f"📂 Base Path: {BASE_PATH}")
print(f"📄 CSV File: {CSV_FILE_PATH}")
print("-" * 60)

# Check if file exists
if os.path.exists(CSV_FILE_PATH):
    print(f"✅ Main data file FOUND!")

    # Load the dataset
    df = pd.read_csv(CSV_FILE_PATH)

    print(f"\n📊 Dataset loaded successfully!")
    print(f"📐 Shape: {df.shape[0]} rows × {df.shape[1]} columns")
    print(f"💾 Memory: {df.memory_usage(deep=True).sum() / 1024**2:.2f} MB")

    print("\n🔍 First 5 rows:")
    display(df.head())

    print("\n📋 Column names:")
    print(df.columns.tolist())

else:
    print(f"❌ File NOT found at: {CSV_FILE_PATH}")
    print("\n📂 Let's explore what's in your folder:")
    if os.path.exists(BASE_PATH):
        for item in os.listdir(BASE_PATH):
            item_path = os.path.join(BASE_PATH, item)
            if os.path.isdir(item_path):
                print(f"   📂 {item}/")
            else:
                print(f"   📄 {item}")

# Check indicators file
if os.path.exists(INDICATORS_PATH):
    print(f"\n✅ Indicators file found!")
else:
    print(f"\n⚠️ Indicators file not found (optional)")
```
